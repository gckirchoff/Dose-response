# Formulas
## Change in susceptible resistant ratio
$\frac{d(Cells_{S/R})}{dt}=(GR_{S/R}-DR_{S/R})*Cells_{S/R}$

This describes the change in ratio of S to R cells/time

```python
def dCellsSR_dt(GR, DR, Cells):
    return (GR-DR)*Cells
```

```python
GR = 23
DR = 14
Cells = 250
dCellsSR_dt(GR, DR, Cells)
```
